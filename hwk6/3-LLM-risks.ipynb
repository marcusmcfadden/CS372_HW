## Task 1

Referencing the article but in your own words, articulate **at least two** societal challenges or risks assosciated with large language models. Explain in 1-2 paragraphs.

Large language models on the surface should provide better Natural Language Processing capabaility; however, in the article it critically examines risks of a trajectory towards "bigger=better" models. Firstly, the article examines environmental impacts, both from an objective analysis and authoritative supremacy. Large models require immense computing power, resulting in large carbon footprints. Specifically, "the authors trained a Transformer (big) model with neural architecture search and estimated that the training procedure emitted 284t of $\text{CO}_2$." And that environmental impact does not just affect a proximal western society, but impacts residents of other nations and cultures, which largely don't even utilize English. So the question the article ends up raising is based in a glimpse of racism: "Is it fair or just to ask... that the residents... affected by drastic floods pay the environmental price of training and deploying ever larger English LMs, when similar large-scale models aren’t being produced for Dhivehi or Sudanese Arabic?"

Secondly, the article examines risk associated with LLM from the perspective of perpetuated structural bias in training data. It examines how those who are more likely to conform to prejudiced idealogy are overrepresented in training data. As such it "set[s] up models trained on these datasets to further amplify biases and harms." Overall the article argues that with an unequitable glimpse of discourse and viewpoints, those of marginalized populations fail to be represented in training data, ultimately leading to a model which fails to interpret marginilized discourse positively.

## Task 2

In Section 7, the article discusses Paths Forward. Briefly summarize the authors' suggestions. Do you believe these suggestions can mitigate the risks you identified in task 1? Why or why not? Explain in 1-2 paragraphs.

The author suggests that the community behind NLP adopt proactive planning instead of reactive fixes as it relates to the risks associated with LLMs. The author suggests factoring in environmental, energy, and financial costs before initiating projects, with efficiency and optimization of such costs a primary evaluation metric behind how much "better" a model is. They advocate for pre-vetting datasets to account for structural biases present as well as integrating Value Sensitive Design to identify and further account for marginalized stakeholder effect. Overall they suggest reform where effort is moved from engineering ever-larger models and towards more efficient, equitable architecture buttressed by regulatroy policy solutions.

I believe that these solutions can mitigate risk; however, only if implemented and utilized properly. So often in modern society is ever-growing technology a spearhead for large coorporations to snub laborers and evolve their architecture on the backs of many. It would need to take proactive, government-orchestrated policy reform at the highest echelons in order to fully implement these processes. Regardless, I do believe engineering models optimized for cost effectiveness and curated datasets could mitigate issues. It is truly when engineers are faced with finite resources and large constraints that the product is manufactured with ingenuity and refinement.
