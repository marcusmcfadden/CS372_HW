We have elected to read article 3 from Hawkins et al.

The article covers the alarming rate at which AI models are being used to generate deepfake images of celebrities/online personalities. As discussed in the article, deepfake models can be rapidly created and distributed on model-sharing sites like Civitai and HuggingFace, despite these models going against the site's Terms of Service. The study focuses on Civtai, and finds that nearly 2,000 developers have uploaded nearly 35,000 deepfake image generation models tagged 'celebrity', 'nsfw', 'Instagram', or similar. 

The article suggests a few courses of action to combat this. One that the article mentions is enforcing the site's own Terms of Service. These sites are able to effectively enforce their terms of service in certain contexts, like preventing the uploads of or removing models generating child sexual-abuse imagery. The article also mentions that a model developer may be required to provide evidence of consent from an individual if a model is meant to generate images about them, or that model-devlopers/platforms may work to add watermarks so that it becomes clear downstream which models were used to create deepfake imagery. The final strategy suggested by the article revolves around regulations: currently the process of sharing deepfake images is highly regulated, but the area of creating models to make these images is a far cloudier subject matter.

It is also worth noting that some of the limitations of the paper may limit the efficacy of their suggested interventions. The interventions suggested by the paper largely revolve around model-sharing sites enforcing their Terms of Service, though the paper acknowledges that these sites do not cover the whole breadth of image generation models. Enforcing Terms of Service can only impact the models available on sites like Civitai. The paper also acknowledges that tracking the number of models on these sites and the number of model downloads does not provide an accurate scale for the number of images that these models have generated. As such, the issue may be far worse than suggested in the paper.
